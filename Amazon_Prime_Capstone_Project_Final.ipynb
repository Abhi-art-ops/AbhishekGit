# **Project Summary -**

Amazon Prime Video is one of the largest streaming platforms in the world, offering thousands of movies
and TV shows across genres, languages, and production countries. As the platform continues to expand its
content library, understanding *what* is being added, *how* it performs, and *who* is contributing to it

This project performs an end-to-end Exploratory Data Analysis (EDA) on the Amazon Prime Video dataset, which
is split across two files: `titles.csv` (9,000+ titles with attributes such as show type, genre, release
year, runtime, age certification, and IMDb/TMDb ratings) and `credits.csv` (~124,000+ actor/director credits
linked to those titles).

The analysis begins with data understanding - inspecting the shape, structure, and data types of both
datasets, identifying missing values, duplicates, and inconsistent entries (for example, `genres` and
`production_countries` are stored as stringified lists and need to be parsed before they can be analyzed).
Appropriate cleaning strategies are applied: numerical gaps (like missing IMDb scores) are handled
conservatively so as not to distort distributions, categorical gaps are flagged rather than silently
dropped, and duplicate title IDs are removed.

Once the data is clean, the analysis follows a structured Univariate -> Bivariate -> Multivariate (UBM) flow.
Univariate analysis looks at the shape of individual variables - how titles are split between movies and TV
shows, how the content library has grown over release years, which genres and age certifications are most
common, and how ratings and runtimes are distributed. Bivariate analysis compares pairs of variables - IMDb
score vs TMDb score, ratings by show type, runtime by show type, and genre popularity over time - to surface
relationships that a single variable can't reveal on its own. Multivariate analysis brings several variables
together, using a correlation heatmap of the numeric fields and combined genre/year trends, and folds in the
credits data to see which actors and directors appear most frequently on the platform.

Throughout, each chart is chosen deliberately (bar charts for categorical comparisons, histograms/KDEs for
distributions, scatter plots for numeric relationships, heatmaps for correlation, and line plots for trends
over time), and every chart is followed by a short note on what it reveals and why it matters.

The final section translates these findings into business language: what content strategists can learn about
genre saturation and gaps, what marketing teams can learn about which genres and certifications resonate,
and what investors can learn about production trends, ratings, and audience engagement over time. The project
closes with a concise summary of the most actionable insights and a discussion of natural next steps, such as
deeper genre-level modelling or a recommendation system, that would be reasonable follow-ups but sit outside
the scope of this EDA.

## 3. **Data Wrangling**

### Data Wrangling Code

````python
# Work on a copy so the originally-loaded data stays intact for reference
titles = titles_df.copy()
credits = credits_df.copy()

# --- Parse stringified list columns into real Python lists ---------------
def parse_list_column(value):
    """Safely convert a stringified list like "['drama', 'comedy']" into an
    actual Python list. Returns an empty list for missing/unparseable values."""
    if pd.isna(value):
        return []
    try:
        parsed = py_ast.literal_eval(value)
        return parsed if isinstance(parsed, list) else []
    except (ValueError, SyntaxError):
        return []

titles['genres_list'] = titles['genres'].apply(parse_list_column)
titles['production_countries_list'] = titles['production_countries'].apply(parse_list_column)

# --- Handle missing values -------------------------------------------------
# age_certification: large number of missing values -> fill with an explicit
# 'Not Rated' label rather than dropping rows, so we don't lose other data.
titles['age_certification'] = titles['age_certification'].fillna('Not Rated')

# seasons only applies to SHOWs; missing values for MOVIEs are expected and
# left as NaN so they aren't mistaken for zero seasons.

# imdb_score / tmdb_score / tmdb_popularity: leave missing as NaN so they are
# correctly excluded from rating-based aggregations (filling with the mean
# would artificially inflate/deflate the true distribution).

# imdb_votes: missing votes likely mean the title had no/very few ratings.
# Fill with 0 so vote-count aggregations are not skewed by NaNs.
titles['imdb_votes'] = titles['imdb_votes'].fillna(0)

# --- Feature engineering ----------------------------------------------------
# Bucket release_year into a decade for trend analysis
titles['decade'] = (titles['release_year'] // 10 * 10).astype('Int64')

# Explode genres so each (title, genre) pair is its own row - useful for
# genre-level aggregation without losing the original one-row-per-title table
titles_genres_exploded = titles.explode('genres_list').rename(columns={'genres_list': 'genre'})
titles_genres_exploded = titles_genres_exploded[titles_genres_exploded['genre'].notna()]

print('Titles shape after wrangling         :', titles.shape)
print('Exploded (title, genre) pairs shape  :', titles_genres_exploded.shape)
titles.head()
````

### What all manipulations have you done and insights you found?

- Parsed the stringified `genres` and `production_countries` columns into real Python lists
  (`genres_list`, `production_countries_list`) so they can be exploded and counted.
- Filled missing `age_certification` values with an explicit `'Not Rated'` label instead of dropping those
  rows, preserving ~40%+ of the dataset that would otherwise be lost.
- Left `imdb_score`, `tmdb_score`, and `tmdb_popularity` as `NaN` where missing, since imputing a rating
  would distort the true rating distribution; these are simply excluded from rating-based aggregations.
- Filled missing `imdb_votes` with `0`, treating a missing vote count as "not yet rated by enough users."
- Created a `decade` column from `release_year` to make long-run trend charts easier to read.
- Created an exploded `titles_genres_exploded` table (one row per title-genre pair) since a title can belong
  to multiple genres - this is required to accurately count genre frequency instead of undercounting titles
  with multiple genres.

## 5. **Statistical / Correlation Check**

````python
from scipy import stats

# Check whether the IMDb score difference between Movies and Shows (seen in Chart 7)
# is statistically significant using an independent two-sample t-test
movie_scores = titles.loc[titles['type'] == 'MOVIE', 'imdb_score'].dropna()
show_scores = titles.loc[titles['type'] == 'SHOW', 'imdb_score'].dropna()

t_stat, p_value = stats.ttest_ind(movie_scores, show_scores, equal_var=False)
print(f"T-statistic: {t_stat:.3f}")
print(f"P-value    : {p_value:.5f}")

if p_value < 0.05:
    print('The difference in average IMDb score between Movies and Shows is statistically significant.')
else:
    print('The difference in average IMDb score between Movies and Shows is NOT statistically significant.')
````

With **t ≈ -35.3** and **p ≈ 3.0e-203** (far below the 0.05 threshold), the IMDb-score gap between movies
and TV shows seen in **Chart - 7** is confirmed to be a real, highly significant difference rather than
sampling noise - TV shows genuinely rate higher on average on this platform.
